# Topic: Multidimensional SDEs and Itô's Lemma

##  Foundation Recall

**Standard Wiener Process**

A standard Wiener process $W_t$ satisfies:

- $W_0=0$, with almost surely continuous paths.
- Independent increments over disjoint intervals.
- $W_t-W_s\sim N(0,t-s)$ for $0\le s<t$.

$$
E[W_t]=0,\quad \operatorname{Var}(W_t)=t,\quad
\operatorname{Cov}(W_s,W_t)=\min(s,t)
$$

For a time step $h$:

$$
\boxed{\Delta W_i=\sqrt h\,\epsilon_i,\qquad \epsilon_i\sim N(0,1)}
$$

**Note:** $\Delta W_i$ already contains the $\sqrt h$ scaling.

**Stochastic Differential Equation (SDE)**

$$
\boxed{dX_t=a(X_t,t)\,dt+b(X_t,t)\,dW_t}
$$

$a$ is the **drift** (systematic change), and $b$ is the **diffusion** (random sensitivity).

**Euler–Maruyama Method**

Discretises the continuous SDE for numerical simulation:

$$
\boxed{X_{i+1}=X_i+a(X_i,t_i)h+b(X_i,t_i)\Delta W_i}
$$

Coefficients are evaluated at the beginning of each step. EM is generally an approximation.

**Geometric Brownian Motion (GBM)**

$$
\boxed{dS_t=\mu S_t\,dt+\sigma S_t\,dW_t}
$$

Both drift and diffusion are proportional to the current stock price.

EM update:

$$
S_{i+1}=S_i(1+\mu h+\sigma\Delta W_i)
$$

**Note:** EM can produce negative prices; Itô's lemma will help derive the exact GBM solution.

## Part 1. Introduction

Previously, we learned how to describe and simulate a stochastic process $X_t$. However, financial applications often require studying a transformed process:

$$
\boxed{Y_t=f(X_t,t)}
$$

Examples include $\log S_t$ (log stock price), $1/X_t$ (inverse exchange rate), and $V(S_t,t)$ (option value).

**Main question:** Given the SDE for $X_t$, how do we derive the SDE for $Y_t$?

Ordinary calculus uses the chain rule, but Brownian motion has non-zero **quadratic variation**, making the ordinary chain rule insufficient.

**Itô's lemma provides the stochastic chain rule.**

This topic also extends scalar SDEs to multiple state variables and noise sources, introducing covariance matrices and multidimensional Itô's lemma.

## Part 2. Multidimensional Wiener Processes & Vector SDEs

**Multidimensional Wiener Process**

A standard $d$-dimensional Wiener process consists of $d$ independent standard Wiener processes:

$$
W_t=(W_t^1,\ldots,W_t^d)^\top
$$

Its increments satisfy:

$$
\boxed{\Delta W_t\sim N_d(0,hI_d)}
$$

Hence:

$$
E[W_t]=0,\qquad \operatorname{Cov}(W_t)=tI_d
$$

**Constructing Correlated Brownian Motion**

To construct a Brownian motion with drift $m$ and covariance rate $\Sigma$, define:

$$
\boxed{V_t=mt+LW_t,\qquad LL^\top=\Sigma}
$$

where $L$ is the **Cholesky factor** of the positive definite matrix $\Sigma$.

Using the covariance transformation rule:

$$
\begin{aligned}
\operatorname{Cov}(V_t)
&=L\operatorname{Cov}(W_t)L^\top\\
&=L(tI_d)L^\top\\
&=\boxed{t\Sigma}
\end{aligned}
$$

**Remember:**
- $W_t$: standard Wiener process, covariance rate $I_d$.
- $V_t$: transformed Brownian motion, covariance rate $\Sigma$.
- $\Sigma$ is the covariance **per unit time**, not necessarily the covariance at time $t$.

**Vector Stochastic Differential Equations**

A general multidimensional SDE is:

$$
\boxed{dX_t=a(X_t,t)\,dt+B(X_t,t)\,dW_t}
$$

| Object | Dimension | Meaning |
|---|---|---|
| $X_t$ | $n\times1$ | State vector |
| $a(X_t,t)$ | $n\times1$ | Drift vector |
| $B(X_t,t)$ | $n\times d$ | Diffusion matrix |
| $W_t$ | $d\times1$ | Independent noise sources |

**State dimension $n$ need not equal noise dimension $d$.**

Each **row of $B$** corresponds to a state variable; each **column** describes how one independent Wiener process affects all state variables.

For example, two state variables may share one Wiener process ($n=2$, $d=1$):

$$
dX_t=
\begin{pmatrix}1\\2\end{pmatrix}dW_t
$$

Thus, $dX_t^1=dW_t$ and $dX_t^2=2dW_t$.

**Multidimensional Euler–Maruyama**

$$
\boxed{
X_{i+1}=X_i+a(X_i,t_i)h_i+B(X_i,t_i)\Delta W_i
}
$$

where:

$$
\Delta W_i\sim N_d(0,h_iI_d)
$$

The random state increment has conditional covariance:

$$
\boxed{
\operatorname{Cov}(B\Delta W_i\mid X_i)
=h_iBB^\top
}
$$

Therefore, $BB^\top$ describes the **instantaneous covariance rate of state changes**.

## Part 3. Multidimensional Geometric Brownian Motion

**Modelling Multiple Stock Prices**

For $n$ stocks driven by $d$ independent Wiener processes, the multidimensional GBM model is:

$$
\boxed{
dS_t=\operatorname{diag}(S_t)\mu\,dt
+\operatorname{diag}(S_t)\sigma\,dW_t
}
$$

where:

- $S_t\in\mathbb R^n$: stock price vector.
- $\mu\in\mathbb R^n$: drift vector.
- $\sigma\in\mathbb R^{n\times d}$: **volatility loading matrix**.
- $W_t\in\mathbb R^d$: independent standard Wiener processes.

Each **row of $\sigma$** corresponds to a stock, and each **column** corresponds to a Wiener process.

The diagonal matrix $\operatorname{diag}(S_t)$ multiplies each stock's proportional changes by its own current price.

In the general vector SDE notation:

$$
a(s,t)=\operatorname{diag}(s)\mu,\qquad
B(s,t)=\operatorname{diag}(s)\sigma
$$

**Covariance of Stock Returns**

Consider the example from the lecture:

$$
\mu=
\begin{pmatrix}
0.03\\
0.05
\end{pmatrix},
\qquad
\sigma=
\begin{pmatrix}
0.2&0\\
0.05&0.05
\end{pmatrix}
$$

Over a short interval $h$, the random parts of the proportional price changes are approximately:

$$
R_1=0.2\Delta W^1,\qquad
R_2=0.05\Delta W^1+0.05\Delta W^2
$$

Both stocks share the noise source $W^1$, so their random returns are correlated even though $W^1$ and $W^2$ are independent.

Using $\operatorname{Cov}(\Delta W)=hI_d$:

$$
\boxed{
\operatorname{Cov}(R)=h\sigma\sigma^\top
=h\begin{pmatrix}
0.04&0.01\\
0.01&0.005
\end{pmatrix}
}
$$

- **Diagonal entries:** Variances of individual stock returns.
- **Off-diagonal entries:** Covariances between stock returns.

**Important distinction:**

$$
\underbrace{\sigma}_{\text{Volatility loading}}
\qquad
\underbrace{\sigma\sigma^\top}_{\text{Covariance rate}}
\qquad
\underbrace{h\sigma\sigma^\top}_{\text{Covariance over time }h}
$$

Here, the covariance refers to the **random proportional price changes**, not the stock price levels.

**Euler–Maruyama Simulation**

Applying the vector Euler–Maruyama method:

$$
\boxed{
S_{i+1}
=S_i+\operatorname{diag}(S_i)\mu h
+\operatorname{diag}(S_i)\sigma\sqrt h\,\epsilon_i
}
$$

where $\epsilon_i\sim N_d(0,I_d)$.

In NumPy, the key operations are:

- `sigma @ epsilon[:, i]`: combines independent noise sources for each stock.
- `np.sqrt(dt)`: scales standard normal shocks into Wiener increments.
- `S[:, i] * (...)`: converts proportional changes into price changes through elementwise multiplication.

## Part 4. From ODE to Stochastic Calculus

**Changing Variables in an ODE**

For a deterministic ODE:

$$
\frac{dx_t}{dt}=g(x_t)
$$

If $y_t=f(x_t)$, the ordinary chain rule gives:

$$
\boxed{\frac{dy_t}{dt}=Df(x_t)g(x_t)}
$$

$Df$ is the **Jacobian matrix**, containing the first partial derivatives of $f$.

For example, $f(u,v)=(2u,v)^\top$ has Jacobian:

$$
Df=
\begin{pmatrix}
2&0\\
0&1
\end{pmatrix}
$$

This transformation stretches the horizontal coordinate by a factor of two.

**Why Squared Brownian Increments Matter**

Consider $f(x)=x^2$ and $X_t=W_t$:

$$
W_{t+h}^2-W_t^2
=2W_t\Delta W_t+(\Delta W_t)^2
$$

For a smooth ODE solution, $\Delta x=O(h)$, so the sum of squared increments vanishes as $h\to0$.

For Brownian motion:

$$
\Delta W\sim N(0,h),\qquad E[(\Delta W)^2]=h
$$

Over $N=T/h$ intervals:

$$
\boxed{
\sum_{i=1}^{N}(\Delta W_i)^2
\xrightarrow{L^2}T
}
$$

This is **quadratic variation**.

In Itô calculus, we use the formal rules:

$$
\boxed{(dW_t)^2=dt,\qquad dt\,dW_t=0,\qquad (dt)^2=0}
$$

**Key insight:** Squared Brownian increments cannot be ignored because their accumulated contribution becomes a time term. This explains the additional drift in Itô's lemma.

**SDEs as Fields of Curves**

An ODE can be visualised using velocity vectors. For an SDE, we can instead attach a curve to each state:

$$
\boxed{\gamma(x,t,s)=x+b(x,t)s+a(x,t)s^2}
$$

Setting $s=\Delta W_t$ gives an alternative numerical update:

$$
X_{t+h}=X_t+b\Delta W_t+a(\Delta W_t)^2
$$

Since squared Wiener increments accumulate like time increments, this curved scheme and Euler–Maruyama converge to the same SDE solution under suitable assumptions.

The coefficients can be recovered from the curve:

$$
\boxed{
\gamma_s(x,t,0)=b(x,t),\qquad
\frac12\gamma_{ss}(x,t,0)=a(x,t)
}
$$

The first derivative represents **diffusion**, while half the second derivative represents **drift**.

This geometric representation helps explain how changing variables produces the Itô correction.

## Part 5. Itô's Lemma

**Scalar Itô's Lemma**

Suppose:

$$
dX_t=a(X_t,t)\,dt+b(X_t,t)\,dW_t
$$

For a sufficiently smooth function $Y_t=f(X_t,t)$:

$$
\boxed{
dY_t=
\left(f_t+af_x+\frac12b^2f_{xx}\right)dt
+bf_x\,dW_t
}
$$

- $f_t$: explicit time dependence.
- $af_x$: drift from the ordinary chain rule.
- $\frac12b^2f_{xx}$: **Itô correction**.
- $bf_x$: transformed diffusion.

All derivatives and coefficients are evaluated at $(X_t,t)$. If $f$ has no explicit time dependence, $f_t=0$.

**Where Does the Itô Correction Come From?**

Using the second-order Taylor expansion:

$$
\Delta f\approx f_t h+f_x\Delta X+\frac12f_{xx}(\Delta X)^2
$$

Since:

$$
\Delta X\approx ah+b\Delta W
$$

and $(\Delta W)^2$ contributes at order $h$:

$$
(\Delta X)^2\rightsquigarrow b^2h
$$

Therefore:

$$
\Delta f\approx
\left(f_t+af_x+\frac12b^2f_{xx}\right)h
+bf_x\Delta W
$$

The extra drift comes from **second-order Taylor expansion and Brownian quadratic variation**.

The curved-scheme derivation gives the same result: applying $f$ to $\gamma(x,t,s)$ and differentiating twice produces the coefficient $af_x+\frac12b^2f_{xx}$.

**Example: Squared Brownian Motion**

Take $X_t=W_t$ and $f(x)=x^2$.

Since $a=0$, $b=1$, $f_x=2x$ and $f_{xx}=2$:

$$
\boxed{d(W_t^2)=dt+2W_t\,dW_t}
$$

The ordinary chain rule would incorrectly omit $dt$.

**Multidimensional Itô's Lemma**

For an $n$-dimensional state driven by $d$ independent Wiener processes:

$$
dX_t=a(X_t,t)\,dt+B(X_t,t)\,dW_t
$$

Define the covariance rate matrix:

$$
C=BB^\top
$$

For a scalar function $f(X_t,t)$:

$$
\boxed{
\begin{aligned}
df(X_t,t)
={}&
\left(
f_t+\sum_{i=1}^{n}a_if_{x_i}
+\frac12\sum_{i=1}^{n}\sum_{j=1}^{n}
C_{ij}f_{x_ix_j}
\right)dt\\
&+\sum_{\alpha=1}^{d}
\left(\sum_{i=1}^{n}f_{x_i}B_{i\alpha}\right)dW_t^\alpha
\end{aligned}
}
$$

The matrix of second derivatives $(f_{x_ix_j})$ is the **Hessian matrix**.

A useful compact form is:

$$
\boxed{
df=
\left(
f_t+\nabla f^\top a+
\frac12\operatorname{tr}(BB^\top H_f)
\right)dt
+\nabla f^\top B\,dW_t
}
$$

**Two-State Formula**

For:

$$
dX=a_1dt+b_1dW^1,\qquad
dY=a_2dt+b_2dW^2
$$

with correlation $\rho$ between the Wiener increments:

$$
\boxed{
\begin{aligned}
df(X,Y,t)
={}&
\left(
f_t+a_1f_x+a_2f_y
+\frac12b_1^2f_{xx}
+\rho b_1b_2f_{xy}
+\frac12b_2^2f_{yy}
\right)dt\\
&+b_1f_x\,dW^1+b_2f_y\,dW^2
\end{aligned}
}
$$

- $\rho=0$: independent Wiener processes.
- $\rho=1$: both equations use the same Wiener process.
- The mixed derivative $f_{xy}$ accounts for covariance between state changes.

**Common Mistakes**

- Do not use the ordinary chain rule for nonlinear transformations of an SDE.
- Include $f_t$ when $f$ explicitly depends on time.
- Remember that $b^2$ appears in the scalar Itô correction.
- In multiple dimensions, use the covariance rate $BB^\top$, not $B$ itself.
- Do not omit mixed derivatives without checking whether they vanish.
- State dimension $n$ and noise dimension $d$ need not be equal.

**Exercise Strategy**

Identify $a$ and $B$, calculate the required derivatives, substitute into Itô's formula, and simplify the resulting drift and diffusion. Where possible, express the final SDE in terms of the transformed process.

## Part 6. Applications of Itô's Lemma

**Solving Geometric Brownian Motion**

Consider:

$$
dS_t=\mu S_t\,dt+\sigma S_t\,dW_t,\qquad S_0>0
$$

Choose the logarithmic transformation:

$$
Y_t=\log S_t
$$

Since $f_s=1/s$ and $f_{ss}=-1/s^2$, Itô's lemma gives:

$$
\boxed{
d\log S_t=
\left(\mu-\frac12\sigma^2\right)dt+\sigma\,dW_t
}
$$

The transformed drift and diffusion are constant, so integration yields:

$$
\log S_t=\log S_0+
\left(\mu-\frac12\sigma^2\right)t+\sigma W_t
$$

Therefore, the **exact GBM solution** is:

$$
\boxed{
S_t=S_0\exp\left[
\left(\mu-\frac12\sigma^2\right)t+\sigma W_t
\right]
}
$$

**Implications:**

- $S_t>0$ whenever $S_0>0$.
- The log stock price is normally distributed.
- Constant-parameter GBM can be simulated exactly at grid times without Euler–Maruyama discretisation error.

For a time step $h$:

$$
\boxed{
S_{t+h}=S_t\exp\left[
\left(\mu-\frac12\sigma^2\right)h+\sigma\Delta W
\right]
}
$$

**Brownian Motion on a Circle**

Let:

$$
\theta_t=W_t,\qquad
U_t=\cos W_t,\qquad V_t=\sin W_t
$$

Applying scalar Itô's lemma to each coordinate:

$$
\boxed{
\begin{aligned}
dU_t&=-\frac12U_t\,dt-V_t\,dW_t\\
dV_t&=-\frac12V_t\,dt+U_t\,dW_t
\end{aligned}
}
$$

To verify that the process remains on the unit circle, define:

$$
f(u,v)=u^2+v^2
$$

Using the two-state Itô formula, with the same Wiener process driving both coordinates:

$$
\text{Drift}
=2u\left(-\frac u2\right)
+2v\left(-\frac v2\right)
+(-v)^2+u^2=0
$$

$$
\text{Diffusion}=2u(-v)+2vu=0
$$

Hence:

$$
\boxed{d(U_t^2+V_t^2)=0}
$$

With initial state $(U_0,V_0)=(1,0)$:

$$
\boxed{U_t^2+V_t^2=1}
$$

**Key insight:** The inward drift compensates for the quadratic effect of tangential noise.

Without this inward drift, the equations

$$
dU=V\,dW,\qquad dV=-U\,dW
$$

satisfy:

$$
d(U^2+V^2)=(U^2+V^2)\,dt
$$

so the radius grows rather than remaining constant.

This shows why **tangential diffusion alone is insufficient to keep an Itô process on a circle**.
